```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
import numpy as np
```

```python
review=[
    ["The book was great",1],
    ["Amazing work and content",1],
    ["Not good",0],
    ["Liked the twist",1],
    ["Disappointing",0]
]

df=pd.DataFrame(review,columns=["review","sentiment"])
```

```python
df["review"]=df["review"].str.lower()
df
```

##### Bag of Words:

```python
cv=CountVectorizer(max_features=5)
```

```python
bow=cv.fit_transform(df["review"])
```

```python
print(cv.vocabulary_)
```

```python
print(bow[0].toarray())
```
