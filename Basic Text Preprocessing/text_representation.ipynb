```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
import numpy as np
```

```python
review=[
    ["The book was great",1],
    ["Amazing work and content",1],
    ["Not good",0],
    ["Liked the twist",1],
    ["Disappointing",0]
]

df=pd.DataFrame(review,columns=["review","sentiment"])
```

```python
df["review"]=df["review"].str.lower()
df
```

##### Bag of Words:

```python
cv=CountVectorizer(max_features=5)
```

```python
bow=cv.fit_transform(df["review"])
```

```python
print(cv.vocabulary_)
```

```python
print(bow[0].toarray())
```

##### N-grams BOW

```python
phone_review=[
    ["Has amazing capability",1],
    ["The speed is very good",1],
    ["Not value for money",0],
    ["Loved it",1]
]
df2=pd.DataFrame(phone_review,columns=["review","sentiment"])
```

```python
df2["review"]=df2["review"].str.lower()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(ngram_range=(1,2))
```

```python
bow=cv.fit_transform(df2["review"])
```

```python
print(cv.vocabulary_)
```

```python
print(bow[0].toarray())
```
