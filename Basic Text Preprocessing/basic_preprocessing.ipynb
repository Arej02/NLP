```python
# 2) Remove punctuation using a build in function:
punc=string.punctuation
print(punc)
```

```python
def remove_punctuation(para):
    for text in para:
        if text in punc:
            para=para.replace(text,'')
    return para

```

```python
text1="My name is Arya. What is your name?"

res=remove_punctuation(text1)
print(res)
```

```python
# 3) Convert the slang words to normal words:
abbreviations={
    "ASAP": "As Soon As Possible",
    "ATM": "At The Moment",
    "BTW": "By The Way",
    "DM": "Direct Message",
    "FAQ": "Frequently Asked Questions",
    "FOMO": "Fear Of Missing Out",
    "FYI": "For Your Information",
    "GN": "Good Night",
    "GOAT": "Greatest Of All Time",
}
```

```python
def remove_slang(text):
    new_text=[]
    for word in text.split():
        if word.upper() in abbreviations:
            new_text.append(abbreviations[word.upper()])
        else:
            new_text.append(word)

    
    return " ".join(new_text)
```

```python
test1="He is GOAT"
res=remove_slang(test1)
print(res)
```

```python
test1="Gn world,FYI, tomorrow is new year so dm asap"
res=remove_slang(test1)
print(res) # Since FYI is tokenized as FYI, it is not converted to it's keys
```
